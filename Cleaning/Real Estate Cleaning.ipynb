```python
import pandas as pd
```

```python
df = pd.read_csv("House Info.csv")
df
```

```python
df.isnull().any()
```

```python
df['Price_per_SqFt'] = round(df['Sale_Price'] / df['SqFt_Living'], 1)
df
```

```python
Luxury_houses = df[df['Bedrooms'] > 6]
df = df[df['Bedrooms'] <=6 ]
Luxury_houses
```

```python
df.groupby(df['Year_Renovated'] > 0)['Sale_Price'].mean()
```

```python
df[df['Sale_Price'] <= 10000]
```

```python
df['Sale_Price'] = df['Sale_Price'].abs()
```

```python
avg_price = df.loc[df['Sale_Price'] > 0, 'Sale_Price'].mean()

df['Sale_Price'] = df['Sale_Price'].replace(0, avg_price)
```

```python
is_two_digit_typo = (df['Bedrooms'] >= 10) & ((df['SqFt_Living'] / df['Bedrooms']) < 180)

df.loc[is_two_digit_typo, 'Bedrooms'] = df.loc[is_two_digit_typo, 'Bedrooms'] - 10

median_3bed_sqft = df.loc[df['Bedrooms'] == 3, 'SqFt_Living'].median()
is_single_digit_suspect = (df['Bedrooms'] > 6) & (df['Bedrooms'] < 10) & (df['SqFt_Living'] < median_3bed_sqft)

suspect_rows = df[is_single_digit_suspect]
print(suspect_rows[['Bedrooms', 'SqFt_Living']])
```

```python
print(df[df['Bedrooms'] > 6][['Bedrooms', 'SqFt_Living']])
```

```python
price_cutoff = df['Sale_Price'].quantile(0.99)
print(price_cutoff)

top_1_percent = df[df['Sale_Price'] > price_cutoff]

print(top_1_percent[['Sale_Price', 'Bedrooms', 'SqFt_Living']])
```

```python
df.to_csv('House Info Updated.csv', index=False)
```
